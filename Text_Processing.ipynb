**Text Preprocessing**

```python
def preprocess_text(text):

    if pd.isna(text):
        return ""

    text = str(text).lower()

    text = text.replace("–", "-")
    text = text.replace("—", "-")
    text = text.replace("−", "-")
    text = text.replace("’", "'")

    text = re.sub(
        r'(\d+)\s+(\d+)\s+(years?|yrs?)',
        r'\1-\2 \3',
        text
    )
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
def preprocess_location(text):

    if pd.isna(text):
        return ""

    text = str(text).lower()
    text = text.strip()
    text = text.replace(".", "")
    text = re.sub(r"\s+", " ", text)

    return text
```

```python
def preprocess_company(text):

    if pd.isna(text):
        return ""

    text = str(text)
    text = re.sub(r"[\n\t]+", " ", text)
    text = re.sub(
        r"\s+\d+\.\d+$",
        "",
        text
    )
    text = re.sub(
        r"\s+\d+$",
        "",
        text
    )
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

**Normalisasi**

```python
def build_pattern(dictionary):
    patterns = {}

    for key, values in dictionary.items():

        vals = sorted(values, key=len, reverse=True)

        escaped = []

        for v in vals:
            escaped.append(re.escape(v))

        patterns[key] = re.compile(
            r'\b(?:' + '|'.join(map(re.escape, vals)) + r')\b'
        )

    return patterns


education_pattern = build_pattern(education_level)
major_pattern = build_pattern(education_major)
career_pattern = build_pattern(career_level)
hard_pattern = build_pattern(hard_skills)
soft_pattern = build_pattern(soft_skills)
```

```python
number_words = {
    "zero":0,
    "one":1,
    "two":2,
    "three":3,
    "four":4,
    "five":5,
    "six":6,
    "seven":7,
    "eight":8,
    "nine":9,
    "ten":10,
    "eleven":11,
    "twelve":12,
    "thirteen":13,
    "fourteen":14,
    "fifteen":15,
    "sixteen":16,
    "seventeen":17,
    "eighteen":18,
    "nineteen":19,
    "twenty":20
}
```

```python
def extract_education(text):
    if pd.isna(text):
        return "Unspecified"

    text = preprocess_text(text)
    result = []

    for edu, keywords in education_level.items():
        pattern = r'\b(' + '|'.join([re.escape(k) for k in keywords]) + r')\b'
        if re.search(pattern, text, flags=re.IGNORECASE):
            result.append(edu)

    return ", ".join(sorted(set(result))) if result else "Unspecified"
```

```python
def extract_major(text):

    if pd.isna(text):
        return None

    text = preprocess_text(text)

    result = []

    for major, pattern in major_pattern.items():

        if pattern.search(text):

            result.append(major)

    return ", ".join(sorted(set(result))) if result else "Other"
```

```python
def extract_hard_skill(text):

    if pd.isna(text):
        return None

    text = preprocess_text(text)

    result = []

    for skill, pattern in hard_pattern.items():

        if pattern.search(text):

            result.append(skill)

    return ", ".join(sorted(set(result))) if result else "Other"
```

```python
def extract_soft_skill(text):

    if pd.isna(text):
        return None

    text = preprocess_text(text)

    result = []

    for skill, pattern in soft_pattern.items():

        if pattern.search(text):

            result.append(skill)

    return ", ".join(sorted(set(result))) if result else "Other"
```

```python
def extract_experience(text):
    if pd.isna(text):
        return None

    text = preprocess_text(text)

    for word, num in number_words.items():
        text = re.sub(rf'\b{word}\b', str(num), text)

    text = re.sub(
        rf'\b(?:over|nearly|about|around|almost|approximately)\s+\d+\s*(?:years?|yrs?)\b(?:\s+ago)?',
        '',
        text
    )
    text = re.sub(
        rf'\b\d+\s*(?:years?|yrs?)\s+ago\b',
        '',
        text
    )

    if re.search(r'fresh\s*graduate|freshers?|no\s*experience|entry\s*level', text):
        return 0

    m = re.search(r'(\d+)\+?\s*(?:or\s+more)?\s*(?:-|to)\s*(\d+)\+?\s*(?:or\s+more)?\s*(?:years?|yrs?)', text)
    if m:
        return int(m.group(1))

    patterns = [
        r'(?:minimum|min\.?|at\s+least|required|requirement|minimum\s+of)\s*(\d+)\+?\s*(?:years?|yrs?)',
        r'(\d+)\+?\s*(?:years?|yrs?)\s*(?:of\s+)?(?:professional\s+)?experience',
        r'experience\s*(?:of\s*)?(\d+)\+?\s*(?:years?|yrs?)'
    ]

    for pattern in patterns:
        m = re.search(pattern, text)
        if m:
            return int(m.group(1))

    for exp, variants in minimum_experience.items():
        for v in variants:
            if preprocess_text(v) in text:
                return int(exp)

    m = re.search(r'(\d+)\+?\s*(?:years?|yrs?)', text)
    if m:
        return int(m.group(1))

    return None
```

```python
def infer_level_from_experience(exp):

    if exp is None:
        return None

    if exp == 0:
        return "Entry"

    elif exp <= 2:
        return "Entry"

    elif exp <= 4:
        return "Mid"

    elif exp <= 7:
        return "Senior"

    elif exp <= 10:
        return "Lead"

    else:
        return "Manager"
```

```python
def extract_level(position, description):

    if pd.notna(position):

        for level, pattern in career_pattern.items():

            if pattern.search(position):
                return level

    exp = extract_experience(description)

    return infer_level_from_experience(exp)
```

```python
def extract_state(location):

    if pd.isna(location):
        return "Unknown"

    location = preprocess_location(location)

    if "remote" in location:
            return "Remote"

    match = re.search(
        r',\s*([a-z]{2})\b',
        location
    )

    if match:

        code = match.group(1)

        if code in state_mapping:
            return state_mapping[code]


    for code, state in state_mapping.items():

        if state.lower() in location:
            return state


    if "united states" in location or "usa" in location:
        return "Unknown"


    return "Unknown"
```

```python
tqdm.pandas()

df_merge["deskripsi_clean"] = df_merge["deskripsi"].progress_apply(
    preprocess_text
)

df_merge["posisi_clean"] = df_merge["posisi"].progress_apply(
    preprocess_text
)

df_merge["pendidikan"] = df_merge["deskripsi_clean"].progress_apply(
    extract_education
)

df_merge["jurusan"] = df_merge["deskripsi_clean"].progress_apply(
    extract_major
)

df_merge["pengalaman"] = df_merge["deskripsi_clean"].progress_apply(
    extract_experience
)

df_merge["hard_skill"] = df_merge["deskripsi_clean"].progress_apply(
    extract_hard_skill
)

df_merge["soft_skill"] = df_merge["deskripsi_clean"].progress_apply(
    extract_soft_skill
)

df_merge["level"] = df_merge.progress_apply(

    lambda row: extract_level(
        row["posisi_clean"],
        row["deskripsi_clean"]
    ),

    axis=1
)

df_merge["lokasi"] = df_merge["lokasi"].progress_apply(
    extract_state
)

df_merge["perusahaan"] = df_merge["perusahaan"].progress_apply(
    preprocess_company
)

df_merge['deskripsi_len'] = df_merge['deskripsi'].fillna("").str.split().apply(len)
```

```python
hard = (
    df_merge[['tahun', 'hard_skill']]
    .dropna()
    .assign(hard_skill=lambda x: x['hard_skill'].str.split(','))
    .explode('hard_skill')
)

hard['hard_skill'] = hard['hard_skill'].str.strip()

hard_count = (
    hard.groupby(['tahun', 'hard_skill'])
        .size()
        .reset_index(name='jumlah')
)

hard_count['persentase'] = (
    hard_count['jumlah']
    / hard_count.groupby('tahun')['jumlah'].transform('sum')
    * 100
).round(2)

top10_hard = (
    hard_count.sort_values(
        ['tahun', 'persentase'],
        ascending=[True, False]
    )
    .groupby('tahun')
    .head(5)
)

print(top10_hard)
```
